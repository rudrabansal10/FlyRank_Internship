## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

The dataset contains one aggregated 90-day snapshot per content item rather than daily history, so it cannot capture temporal patterns within that window.

Missing keyword-related fields are systematic and depend on content_type, meaning they are not missing at random.

The dataset is anonymized, so client identities, URLs, and search queries are unavailable.

Observational data can reveal associations but cannot establish causal relationships.

trend_direction and trend_pct define the target and therefore cannot be used as model features without causing data leakage.
